#### What types of model or models do you plan to use and why?

The project will begin with a Multiple Linear Regression model to establish a baseline for performance. Beyond linear models, I will explore more flexible machine learning techniques such as Random Forest Regression. #### How do you plan to evaluate your results?

The models will be evaluated using two main metrics: Root Mean Square Error (RMSE) and R² (Coefficient of Determination).

* RMSE will serve as the primary evaluation metric. It provides an interpretable measure of prediction error, showing on average how far off a model’s prediction is from the actual attendance value. Since attendance numbers can vary widely across games, minimizing RMSE is key for producing a practically useful model.
    
* R² score will indicate the proportion of variance in the attendance figures explained by the features in the model. A higher R² suggests a better model fit and helps validate that our predictors are capturing the underlying trends.

To ensure robust results, I will use k-fold cross-validation, likely with five or ten folds, depending on the size and balance of the data. This technique will help avoid overfitting by providing insight into how well the model generalizes to unseen data. I will also maintain a train-test split to assess out-of-sample performance and to compare model results consistently.
	
Feature selection and transformation (e.g., dummy encoding categorical variables such as weather condition or opponent) will be standardized across models. If one model substantially outperforms the others in both RMSE and R², it will be selected as the final model for deployment.

A team's success can influence the amount of people who attend their games. To measure the team's success I'll calculate their winning percentage for each day for each of the ten seasons in our data set.

```python
# Constants
API_KEY = '83d90d908b655cba07bedc8b01371be1'  
LAT = 39.0974  
LON = -84.5071
BASE_URL = 'https://api.openweathermap.org/data/3.0/onecall/timemachine'

home_games['Full_Date'] = pd.to_datetime(home_games['Full_Date'])

# Create columns for weather info
home_games['temp'] = None
home_games['humidity'] = None
home_games['weather_description'] = None

# Query OpenWeatherMap
def get_weather_for_date(timestamp):
    params = {
        'lat': LAT,
        'lon': LON,
        'dt': timestamp,
        'appid': API_KEY,
        'units': 'imperial'
    }
    response = requests.get(BASE_URL, params=params)
    response.raise_for_status()
    return response.json()

for idx, row in home_games.iterrows():
    unix_timestamp = int(row['Full_Date'].timestamp())
    try:
        weather_json = get_weather_for_date(unix_timestamp)
        hourly = weather_json.get('data', weather_json.get('hourly', []))
        if hourly:
           temps = [h.get('temp') for h in hourly if h.get('temp') is not None]
           humidities = [h.get('humidity') for h in hourly if h.get('humidity') is not None]
           descriptions = [h.get('weather', [{}])[0].get('description') for h in hourly]

           home_games.at[idx, 'temp'] = sum(temps) / len(temps) if temps else None
           home_games.at[idx, 'humidity'] = sum(humidities) / len(humidities) if humidities else None
           home_games.at[idx, 'weather_description'] = max(set(descriptions), key=descriptions.count) if descriptions else None

        time.sleep(1)  # Respect API rate limits
    except Exception as e:
        print(f"Error on {row['Full_Date']}: {e}")

print(home_games)
```

To obtain weather conditions at Great American Ballpark I used the above funtion to query openweathermap.org and provided the dates of each home game. The temperature and humidity added to the data frame are averages of those values over the day of the game.

```python
def convert_streak(val):
    if isinstance(val, str):
        if set(val) == {'+'}:
            return len(val)
        elif set(val) == {'-'}:
            return -len(val)
    return 0  

home_games['streak_int'] = home_games['Streak'].apply(convert_streak)
```

In our original data set the team's winning/losing streak is represented by plus (+) and minus (-) signs. The above function converts those symbols into integers that can be used in our model.
```python
# Define target and features
y = home_games_1['Attendance']
X = home_games_1.drop(columns=['Attendance'])

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=55)

# Train linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Predict on test set
y_pred = model.predict(X_test)

# Calculate Root Mean Square Error (RMSE)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = model.score(X_test, y_test)

rmse, r2
```

```python
# Calculate residuals
residuals = y_test - y_pred

# Plot residuals
plt.figure(figsize=(10, 6))
sns.histplot(residuals, bins=30, kde=True)
plt.title("Distribution of Residuals")
plt.xlabel("Residual (Actual - Predicted Attendance)")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
